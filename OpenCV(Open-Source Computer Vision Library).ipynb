#<font color=purple>**How to write a program in OpenCV**  
<font color=purple>**Following are the steps:**</font>
* <font color=red>**Import the required libraries**  
* <font color=red>**Import matplotlib to display images**  
* <font color=red>**Import numpy to store the pixel values in the NumPy array**

```python
#Display a 2D image using the imread() function of cv2.Thus,the complete code that appears
import cv2
from matplotlib import pyplot as plt
import numpy as np
img=cv2.imread('rose.jpg')
plt.imshow(img)
plt.title('Flowers')
plt.axis('off')
plt.show()
```

```python
#Display a 2D image using the imread() function of cv2.Thus,the complete code that appears
import cv2
from matplotlib import pyplot as plt
import numpy as np
img=cv2.imread('parrot.jpg')
plt.imshow(img)
plt.title('bird')
plt.axis('off')
plt.show()
```

Notice the colour of the image ,its because the opencv presents images in BGR(Blue,Green,Red) format ,but now we will convert it into RGB(red,green,blue)as follows  
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))

```python
#Display a 2D image using the imread() function of cv2.Thus,the complete code that appears
import cv2
from matplotlib import pyplot as plt
import numpy as np
img=cv2.imread('parrot.jpg')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Flowers')
plt.axis('off')
plt.show()
```

```python
#Display a 2D image using the imread() function of cv2.Thus,the complete code that appears
import cv2
from matplotlib import pyplot as plt
import numpy as np
img=cv2.imread('rose.jpg')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Flowers')
plt.axis('off')
plt.show()
```

#<font color=purple>To print the image in grayscale

```python
#Display a 2D image using the imread() function of cv2.Thus,the complete code that appears
import cv2
from matplotlib import pyplot as plt
import numpy as np
img=cv2.imread('rose.jpg',0)
plt.imshow(img, cmap='gray')
plt.title('Flowers')
plt.axis('off')
plt.show()
```
