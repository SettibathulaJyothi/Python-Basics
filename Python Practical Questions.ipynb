<font color=blue>**12.To calculate Surface Area and Volume of a Cuboid**</font>

```python
Length=int(input("Enter length : "))
Breadth=int(input("Enter breadth : "))
Height=int(input("Enter height : "))
Volume=Length*Breadth*Height
print("Volume of the Cuboid is: ",Volume)
Surface_Area=2*((Length*Breadth)+(Breadth*Height)+(Height*Length))
print("Surface Area of the Cuboid is: ",Surface_Area)
```

<font color=blue>**18.To check the grade of the student**

```python
marks=float(input("Enter your marks out of hundered:\n "))
if 90<=marks<=100:
  print("Grade A")
elif 75<=marks<=89.9:
  print("Grade B")
elif 60<=marks<=74.9:
  print("Grade C")
elif 45<=marks<=59.5:
  print("Grade D")
elif 33<=marks<=44.9:
  print("Grade E")
elif 0<=marks<=32.9:
  print("Grade F")
else:
  print("Invalid marks")
```
